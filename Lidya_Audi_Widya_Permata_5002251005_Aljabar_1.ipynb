## Teori Himpunan Aksiomatik

Teori Himpunan Aksiomatik adalah cara membangun teori himpuunan berdasarkan sejumlah aturan dasar (aksioma) yang dianggap benar sejak awal, kemudian dari aksioma tersebut kita menurunkan atau membuktikan pernyataan-pernyataan lain.

Secara umum terdapat aksioma-aksioma seperti:
1. Aksioma Ekstensionalitas

 \(A=B\) jika anggota \(A\) dan \(B\)

2. Aksioma Himpunan Kosong

Menjamin adanya \(∅\)

3. Aksioma Pasangan

Untuk dua objek a, b, terdapat himpunan {a,b}

4. Aksioma Gabungan

Jika A adalah himpunan yang anggotanya merupakan himpunan-himpunan, maka terdapat himpunan yang menggabungkan anggota dari himpunan-himpunan tersebut. Secara sederhana:

⋃A

5. Aksioma Himpunan Kuasa

Untuk setiap himpunan \(A\), terdapat:

$$ \mathcal P(A) $$

yang berisi seluruh subhimpunan \(A\).

6. Aksioma Tak Hingga

Menjamin adanya setidaknya satu himpunan tak hingga.

Salah satu contohnya berkaitan dengan:

N = {0,1,2,3,…}

7. Aksioma Pemisahan/Separation

Memungkinkan kita mengambil bagian tertentu dari sebuah himpunan berdasarkan suatu kondisi.

Misalnya dari:

$$ A=\{1,2,3,4,5,6\} $$

kita ingin mengambil bilangan genap:

$$ \{x\in A\mid x\text{ genap}\} $$

hasilnya:

{2,4,6}

8. Aksioma Penggantian/Replacement

Secara sederhana, aksioma ini memungkinkan pembentukan himpunan baru dengan memetakan setiap anggota suatu himpunan melalui aturan tertentu.

9. Aksioma Fondasi/Regularity

Mengatur struktur keanggotaan himpunan agar tidak muncul struktur keanggotaan yang bersifat melingkar/bermasalah.

10. Aksioma Pilihan

Axiom of Choice (AC) menyatakan, secara informal, bahwa dari kumpulan himpunan tak kosong kita dapat memilih satu anggota dari masing-masing himpunan tersebut.

## Implementasi Konsep Teori Himpunan Aksiomatik pada Data Penerimaan Bantuan Pendidikan KIP-K

Data mahasiswa dapat memiliki berbagai karakteristik, seperti kondisi sosial ekonomi, status penerimaan bantuan pendidikan, maupun kategori tertentu berdasarkan hasil pendataan. Dengan jumlah data yang besar, diperlukan suatu cara untuk mengelompokkan dan menganalisis data tersebut secara sistematis agar informasi yang diperoleh dapat lebih mudah dipahami.

Salah satu konsep matematika yang dapat digunakan untuk melakukan pengelompokan dan analisis data adalah teori himpunan. Dalam teori himpunan aksiomatik, suatu himpunan dibangun berdasarkan aturan-aturan dasar yang disebut aksioma, sehingga hubungan antarobjek di dalamnya dapat didefinisikan dan dibuktikan secara sistematis. Konsep seperti irisan, gabungan, selisih, komplemen, dan subset dapat digunakan untuk menggambarkan hubungan antara beberapa kelompok data.

**Latar Belakang Masalah**

Dalam penyaluran bantuan pendidikan, mahasiswa dapat memiliki karakteristik yang berbeda-beda. Misalnya, terdapat mahasiswa yang termasuk kelompok sosial ekonomi tertentu dan mahasiswa yang telah menerima bantuan pendidikan seperti KIP-Kuliah.

Agar proses pendataan dapat dilakukan secara sistematis, mahasiswa dapat dikelompokkan ke dalam beberapa himpunan berdasarkan kriteria tertentu.

Misalkan terdapat 100 mahasiswa yang sedang dilakukan verifikasi.

Diketahui:

a. 60 mahasiswa termasuk kelompok desil 1-4

b. 45 mahasiswa merupakan penerima KIP-K

c. 30 mahasiswa termasuk desil 1-4 dan penerima KIP-K

Yang akan dicari:
1. Berapa mahasiswa yang termasuk kedua kelompok?

2. Berapa mahasiswa yang termasuk minimal salah satu kelompok?

3. Berapa mahasiswa yang termasuk desil 1-4 tetapi bukan penerima KIP-K?

4. Berapa mahasiswa yang tidak termasuk kedua kelompok?

5. Bagaimana kasus tersebut dikaitkan dengan teori himpunan aksiomatik?

**Membentuk Himpunan**

Misalkan:

$$ U=\{\text{seluruh mahasiswa yang diverifikasi}\} $$

Karena terdapat 100 mahasiswa:

$$ |U|=100 $$

Kemudian kita bentuk:

$$ A=\{\text{mahasiswa yang termasuk desil 1--4}\} $$

sehingga:

$$ |A|=60 $$

dan:

B = {mahasiswa penerima KIP Kuliah}

sehingga:

$$ |B|=45 $$

Diketahui:

∣A∩B∣=30

**Hubungannya dengan aksioma Skema Penyisihan**

Kita mempunyai himpunan semesta:

$$ U $$

yang berisi seluruh mahasiswa.

Kemudian kita ingin mengambil anggota-anggota \(U\) yang memenuhi suatu kriteria tertentu.

Untuk kelompok \(A\):

$$ A=\{x\in U\mid x\text{ termasuk desil 1--4}\} $$

Untuk kelompok \(B\):

$$ B=\{x\in U\mid x\text{ merupakan penerima KIP Kuliah}\} $$

Secara konsep, sedang memisahkan/mengambil sebagian anggota dari \(U\) berdasarkan suatu kondisi.

Hal tersebut berkaitan dengan Aksioma Skema Pemisahan (Separation):

$$ \boxed{ A=\{x\in U\mid P(x)\} } $$

Jadi:

Dari seluruh mahasiswa dalam \(U\), kita membentuk kelompok baru berdasarkan sifat atau kriteria tertentu.

**Operasi Himpunan**

Setelah \(A\) dan \(B\) terbentuk, kita menggunakan:

$$ A\cap B $$ $$ A\cup B $$ $$ A-B $$ $$ (A\cup B)^c $$

untuk menganalisis data.

Konsep seperti irisan, gabungan, selisih, komplemen, dan subset dapat digunakan untuk menggambarkan hubungan antara beberapa kelompok data.

**Menentukan mahasiswa yang termasuk kedua kelompok**

Dari soal sudah diketahui:

$$ |A\cap B|=30 $$

Jadi:

$$ \boxed{30\text{ mahasiswa}} $$

Interpretasinya:

Terdapat 30 mahasiswa yang memenuhi kedua kriteria sekaligus.

**Menentukan mahasiswa yang termasuk minimal salah satu kelompok**

Pertanyaan:

Berapa mahasiswa yang termasuk desil 1–4 atau merupakan penerima KIP Kuliah?

Kata "atau" menunjukkan gabungan:

$$ A\cup B $$

Gunakan rumus kardinalitas:

$$ |A\cup B| = |A|+|B|-|A\cap B| $$

Substitusi:

$$ |A\cup B| = 60+45-30 $$ $$ =75 $$

Jadi:

∣A∪B∣=75
	Kenapa dikurangi 30?

Karena 30 mahasiswa yang berada di:

$$ A\cap B $$

terhitung dua kali ketika kita menjumlahkan:

$$ 60+45 $$

Jadi harus dikurangi sekali.

**Menentukan mahasiswa yang hanya termasuk desil 1-4**

Pertanyaan:

Berapa mahasiswa yang termasuk desil 1–4 tetapi bukan penerima KIP?

Kita menggunakan selisih:

$$ A-B $$

Rumusnya:

$$ |A-B|=|A|-|A\cap B| $$

Sehingga:

$$ |A-B|=60-30 $$ $$ \boxed{|A-B|=30} $$

Jadi ada:

30 mahasiswa yang termasuk desil 1–4 tetapi bukan penerima KIP.

**Menentukan mahasiswa yang hanya penerima KIP-K**

Sekarang kebalikannya:

$$ B-A $$ $$ |B-A| = |B|-|A\cap B| $$ $$ =45-30 $$ $$ \boxed{15} $$

Jadi ada 15 mahasiswa yang merupakan penerima KIP tetapi tidak termasuk kelompok desil 1–4.

**Menentukan mahasiswa yang tidak termasuk kedua kelompok**

Kita sudah mendapatkan:

$$ |A\cup B|=75 $$

Sedangkan seluruh mahasiswa:

$$ |U|=100 $$

Maka mahasiswa yang tidak berada dalam \(A\cup B\) adalah:

$$ (A\cup B)^c $$

Jumlahnya:

$$ |(A\cup B)^c| = |U|-|A\cup B| $$ $$ =100-75 $$ $$ \boxed{25} $$

Jadi ada 25 mahasiswa yang tidak termasuk desil 1–4 maupun penerima KIP.

**Aksioma Skema Pemisahan**

Dari:

$$ U=\{\text{seluruh mahasiswa}\} $$

kita membentuk:

$$ A=\{x\in U\mid P(x)\} $$

dan:

$$ B=\{x\in U\mid Q(x)\} $$

berdasarkan kriteria tertentu.

	​
**Contoh sederhana secara komputasi**



```python
# Data seluruh mahasiswa
U = {
    "M01", "M02", "M03", "M04", "M05",
    "M06", "M07", "M08", "M09", "M10"
}

# Mahasiswa yang termasuk desil 1-4
A = {
    "M01", "M02", "M03", "M04", "M05"
}

# Mahasiswa penerima KIP Kuliah
B = {
    "M03", "M04", "M05", "M06", "M07"
}

# Irisan
irisan = A & B

# Gabungan
gabungan = A | B

# Selisih A - B
selisih_A_B = A - B

# Selisih B - A
selisih_B_A = B - A

# Komplemen dari gabungan terhadap U
komplemen = U - gabungan

print("Mahasiswa desil 1-4:", A)
print("Mahasiswa penerima KIP:", B)
print("A ∩ B:", irisan)
print("A ∪ B:", gabungan)
print("A - B:", selisih_A_B)
print("B - A:", selisih_B_A)
print("(A ∪ B)^c:", komplemen)
```
